# **2. Referencial Teórico**


Este capítulo tem como objetivo apresentar os fundamentos teóricos e conceituais que sustentam o projeto de previsão das exportações brasileiras no período de 2014 a 2024. Para isso, são abordados os principais conceitos de análise de séries temporais, os modelos clássicos de previsão (ARIMA e SARIMA) e as abordagens modernas baseadas em aprendizado de máquina, com ênfase no algoritmo XGBoost. A integração entre econometria tradicional e técnicas de Machine Learning permite uma visão abrangente e preditiva sobre o comportamento do comércio exterior brasileiro.

**2.1. Análise de Séries Temporais**

Uma série temporal é uma sequência de observações coletadas em intervalos regulares de tempo, cuja análise visa compreender seus padrões de tendência, sazonalidade e ruído (Hyndman & Athanasopoulos, 2018). A tendência representa o movimento de longo prazo; a sazonalidade, os padrões que se repetem periodicamente; e o ruído expressa variações aleatórias sem padrão definido (Box et al., 2015).
No contexto das exportações brasileiras, esses componentes refletem tanto fatores internos (produção agrícola, política cambial) quanto externos (demanda global, crises financeiras e oscilações de commodities).

**2.2. Modelos Clássicos de Previsão: ARIMA e SARIMA**

O modelo ARIMA (AutoRegressive Integrated Moving Average), proposto por Box e Jenkins (1976), é amplamente utilizado para modelar séries temporais univariadas, combinando componentes autorregressivos, diferenciação e médias móveis. Sua formulação permite capturar dependências lineares e tendências ao longo do tempo. Contudo, quando há padrões sazonais regulares, o modelo SARIMA (Seasonal ARIMA) torna-se mais apropriado, pois incorpora termos adicionais para modelar a sazonalidade (Chatfield, 2003).

De acordo com Shumway e Stoffer (2017), o SARIMA é especialmente eficaz em séries econômicas e financeiras que apresentam flutuações periódicas — como é o caso das exportações, que sofrem influência de ciclos agrícolas e variações cambiais. Assim, a aplicação do SARIMA neste projeto justifica-se pela presença de padrões sazonais bem definidos nos dados mensais de exportação do Brasil.

**2.3. Modelos de Aprendizado de Máquina: o Algoritmo XGBoost**

Com o avanço da capacidade computacional, métodos de Machine Learning vêm sendo amplamente aplicados na previsão de séries temporais, oferecendo vantagens na captura de relações não lineares e complexas (Raschka & Mirjalili, 2020).
Entre esses métodos, o XGBoost (eXtreme Gradient Boosting), desenvolvido por Chen e Guestrin (2016), é uma implementação otimizada do Gradient Boosting, capaz de lidar com grandes volumes de dados, controlar o overfitting por meio de regularização (L1 e L2) e incorporar variáveis exógenas relevantes — como preços de commodities ou taxas de câmbio.

Segundo Brownlee (2020), para aplicar o XGBoost a séries temporais, é necessário transformar os dados sequenciais em um formato supervisionado, criando variáveis derivadas de valores passados (lags) e indicadores de tempo (mês, trimestre, ano). Essa abordagem possibilita capturar relações complexas que métodos lineares, como o ARIMA, não conseguem representar adequadamente.

**2.4. Trabalhos Correlacionados e Alternativas de Solução**

Diversos estudos aplicaram ARIMA, SARIMA e XGBoost em problemas semelhantes de previsão econômica.
Silva e Almeida (2021) modelaram as exportações de soja brasileira com SARIMA, obtendo alta acurácia (85%) e confirmando a relevância da sazonalidade anual. Apesar do bom desempenho, os autores destacam limitações em capturar eventos atípicos, como variações cambiais abruptas.

Santos et al. (2022) compararam ARIMA e XGBoost para prever exportações de minério de ferro. O XGBoost apresentou erro médio absoluto (MAE) 12% menor, por incorporar variáveis exógenas como PIB chinês e preço do minério, demonstrando sua superioridade em cenários com múltiplos fatores externos.

Rodrigues e Costa (2023) desenvolveram um modelo híbrido SARIMA-XGBoost, que combinou previsões lineares e não lineares, alcançando desempenho 10% superior aos modelos individuais. Contudo, ressaltaram que o custo computacional e a complexidade de parametrização são maiores.

Li et al. (2020) também evidenciaram a eficácia do XGBoost em previsões de comércio internacional, destacando sua estabilidade mesmo em períodos de volatilidade, embora sua interpretabilidade seja inferior à dos modelos estatísticos tradicionais.

Esses estudos mostram que, enquanto modelos como ARIMA e SARIMA são interpretáveis e eficientes para séries com padrões regulares, algoritmos como o XGBoost são mais adequados para dados complexos e com múltiplas variáveis influenciadoras. Assim, a integração de ambos representa uma alternativa robusta e equilibrada para a previsão das exportações brasileiras.

```python
#Dicionário de países

caminho_pasta = '/content/drive/MyDrive/Mackenzie/Projeto Aplicado IV - Grupo 2'

df_paises = pd.read_csv(
    f'{caminho_pasta}/PAIS.csv',
    encoding='latin1',
    sep=';',
    dtype={'CO_PAIS': str, 'NO_PAIS': str}
)

df_paises['NO_PAIS'] = df_paises['NO_PAIS'].str.replace(' ', '_')

mapa_paises = dict(zip(df_paises['CO_PAIS'], df_paises['NO_PAIS']))

# Substitui os códigos pelos nomes, mantendo o valor original se não houver correspondência
df['CO_PAIS'] = df['CO_PAIS'].map(mapa_paises).fillna(df['CO_PAIS'])

print("\nApós substituição:")
print(df.head())

```

```python
df = df.sort_values('data')

# Remover países com > 40% de nulos nos últimos 2 anos
limite_data = df['data'].max() - pd.DateOffset(months=24)
paises_validos = []
for pais, grupo in df.groupby('CO_PAIS'):
    serie = grupo.groupby('data')['KG_LIQUIDO'].sum()
    idx = pd.date_range(start=limite_data, end=df['data'].max(), freq='MS')
    serie = serie.reindex(idx)

    prop_nulos = serie.isna().mean()
    if prop_nulos <= 0.4:
        paises_validos.append(pais)

df = df[df['CO_PAIS'].isin(paises_validos)]
```

```python
#Criar séries temporais por país

df['data'] = pd.to_datetime(df['data'])

df = df.set_index('data')

# Agrupa por país e agrega por mês (MS = início do mês)
df_mensal = (
    df.groupby('CO_PAIS')['KG_LIQUIDO']
      .resample('MS')
      .sum()
      .reset_index()
)

# Converte o resultado para um dicionário no mesmo formato que antes
series_por_pais = {
    pais: grupo.set_index('data')['KG_LIQUIDO']
    for pais, grupo in df_mensal.groupby('CO_PAIS')
}

print("Séries mensais por país criadas com sucesso!")
```

```python
def tratar_nulos_outliers(serie):
    # Preencher nulos com média móvel central de 3 períodos
    serie = serie.fillna(serie.rolling(window=3, min_periods=1, center=True).mean())

    # Forward/backward fill e fallback final para 0
    serie = serie.ffill().bfill().fillna(0)

    # Tratar outliers usando IQR
    Q1 = serie.quantile(0.25)
    Q3 = serie.quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    # Média móvel para substituir valores fora do intervalo
    media_movel = serie.rolling(window=3, min_periods=1, center=True).mean()
    serie = serie.where((serie >= limite_inferior) & (serie <= limite_superior), media_movel)

    return serie.astype(float)

df['KG_LIQUIDO'] = tratar_nulos_outliers(df['KG_LIQUIDO'])
```

# **5. Modelo base**

O modelo base teve como objetivo construir uma linha de referência sólida para a análise e previsão das exportações mensais (em quilogramas líquidos) por país, entre 2014 e 2025. Foram aplicadas duas abordagens complementares de modelagem: a Regressão Linear, para estimar tendências globais de crescimento, e o modelo SARIMA, voltado para a captura de componentes sazonais e autocorrelacionados das séries temporais.

**5.1 Regressão Linear**

A Regressão Linear foi utilizada como primeiro método de modelagem, com o intuito de identificar a tendência global das exportações ao longo do tempo. Para cada país, ajustou-se um modelo da forma:

                                              Yt = B- + B1t+et
	​
 em que yt representa o volume exportado no tempo T1 e B1 corresponde à inclinação da tendência. Essa abordagem permitiu estimar a variação média mensal das exportações e identificar os países com maior taxa de crescimento recente.

Os resultados mostraram que os cinco países com maior crescimento no último ano foram: China, Uruguai, Índia, Turquia e Estados Unidos. As regressões dessas séries apresentaram coeficientes angulares positivos e expressivos, indicando tendências ascendentes consolidadas, especialmente nas economias de maior volume exportador, como China e Estados Unidos.

A análise de decomposição sazonal revelou ainda padrões cíclicos evidentes, sugerindo a necessidade de modelos capazes de capturar sazonalidade e ruído temporal, o que motivou a aplicação do modelo SARIMA na sequência.

**5.2 Modelo SARIMA**

O modelo SARIMA (Seasonal AutoRegressive Integrated Moving Average) foi empregado para refinar a previsão, incorporando componentes sazonais e autorregressivos. O modelo foi configurado com periodicidade mensal (s = 12) e ajustado às séries suavizadas dos cinco países com maior tendência de crescimento, conforme identificado na etapa anterior.

A suavização — obtida por meio de médias móveis e decomposição sazonal — teve papel crucial na redução do ruído e no aumento da estabilidade das previsões. Após a aplicação do SARIMA, observou-se melhora significativa no desempenho do modelo, especialmente para países com séries temporais longas e volumes de exportação elevados.

Os resultados obtidos são resumidos a seguir:

image.png

Esses resultados demonstram que o SARIMA apresenta excelente desempenho para séries de maior volume e estabilidade, alcançando MAPE inferior a 10% para China, Estados Unidos e Índia — valores considerados muito bons em contextos de previsão econômica.

Por outro lado, países com séries pequenas, incompletas ou altamente voláteis, como Turquia e Uruguai, mantiveram erros mais elevados, refletindo limitações estruturais nos dados e a menor capacidade do modelo em capturar flutuações abruptas.


**5.3 Análise Geral dos Resultados**

De modo geral, a combinação da Regressão Linear e do SARIMA resultou em um modelo base consistente e informativo. A regressão linear se mostrou eficaz para compreender a tendência global de longo prazo, enquanto o SARIMA foi mais adequado para a previsão de curto prazo e para séries com comportamento sazonal marcante.

A aplicação conjunta das duas abordagens permitiu:

- Reduzir o ruído das séries temporais por meio da suavização prévia;

- Melhorar a acurácia preditiva, especialmente em economias de grande escala;

- Evidenciar as limitações em séries com baixa granularidade ou alta variabilidade.

Esses achados consolidam o modelo base como um ponto de partida robusto para as próximas fases do projeto, nas quais poderão ser aplicadas estratégias de otimização de hiperparâmetros, validação cruzada e modelos híbridos de previsão, visando elevar a precisão e a generalização dos resultados obtidos.


```python
#SARIMA para identificaçao dos top 5

import warnings
from statsmodels.tsa.statespace.sarimax import SARIMAX
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")  # evita mensagens de convergência

# Lista para armazenar resultados de previsão
previsoes_sarima = {}

plt.figure(figsize=(14, 7))

for pais in top5['pais']:
    # Usa a série suavizada
    serie = series_suavizadas[pais].sort_index().dropna()

    # Ajuste SARIMA simples: (1,1,1)x(1,1,1,12) como exemplo
    modelo_sarima = SARIMAX(
        serie,
        order=(1,1,1),
        seasonal_order=(1,1,1,12),
        enforce_stationarity=False,
        enforce_invertibility=False
    )
    resultado = modelo_sarima.fit(disp=False)

    # Previsão para os próximos 12 meses
    previsao = resultado.get_forecast(steps=12)
    previsoes_sarima[pais] = previsao

    # Plot série histórica suavizada + previsão SARIMA
    plt.plot(serie.index, serie.values, label=f"{pais} histórico suavizado")
    plt.plot(previsao.predicted_mean.index, previsao.predicted_mean.values, '--', label=f"{pais} previsão SARIMA")

    # Intervalo de confiança da previsão
    intervalo_conf = previsao.conf_int()
    plt.fill_between(intervalo_conf.index,
                     intervalo_conf.iloc[:,0],
                     intervalo_conf.iloc[:,1], alpha=0.2)

plt.title('Top 5 Países - Série Suavizada e Previsão SARIMA')
plt.xlabel('Data')
plt.ylabel('KG Líquido Exportado')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
